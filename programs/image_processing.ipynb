# **Thresholding and Adaptive thresholding**

```python
import cv2 as cv

img = cv.imread('park.jpg',0)
# thres,img_th = cv.threshold(img,130,255,cv.THRESH_BINARY)
# thres,img_th = cv.threshold(img,130,255,cv.THRESH_BINARY_INV)
# thres,img_th = cv.threshold(img,130,None,cv.THRESH_TRUNC)
# thres,img_th = cv.threshold(img,130,None,cv.THRESH_TOZERO)
# thres,img_th = cv.threshold(img,130,None,cv.THRESH_TOZERO_INV)
# thres,img_th = cv.threshold(img,None,255,cv.THRESH_OTSU)
# thres,img_th = cv.threshold(img,None,255,cv.THRESH_TRIANGLE)


adaptive = cv.adaptiveThreshold(img,255,cv.ADAPTIVE_THRESH_GAUSSIAN_C,cv.THRESH_BINARY,11,2.0)

cv.imshow('img',adaptive)
# print(thres)
cv.waitKey(0)
cv.destroyAllWindows()
```

# **HSV, H-hue(degree),s-saturation(percentile),v-value(percentile)**

![hsv](hsv.jpg)

```python
import numpy as np

hue = 120
saturation = 255
value = 255

img_hsv_color = np.uint8([[[hue,saturation,value]]])

img_bgr_color = cv.cvtColor(img_hsv_color,cv.COLOR_HSV2BGR)

print('h',img_hsv_color)
print('b',img_bgr_color)
```

# **LAB L-Lightness(0-100),A - represents Green-red axis(-128-127),B- represents blue-yellow axis(-128-127)**
![lab](lab.jpg)

```python
import cv2 as cv

img = cv.imread('park.jpg')

img_l = cv.cvtColor(img,cv.COLOR_BGR2LAB)

cv.imshow('img',img_l)

cv.waitKey(0)
cv.destroyAllWindows()
```

# **Equalized Histogram**

```python
img = cv.imread('park.jpg',0)
eqh = cv.equalizeHist(img)

cv.imshow('eh',eqh)

cv.waitKey(0)
cv.destroyAllWindows()
```

# **CLAHE(contrast limited adaptive histogram equalization)**

```python
clahe = cv.createCLAHE(clipLimit=2.0,tileGridSize=(8,8))
clahe_img = clahe.apply(img)
cv.imshow('eh',clahe_img)

cv.waitKey(0)
cv.destroyAllWindows()
```
